```python
import numpy as np

X = np.array([[0,0],[0,1],[1,0],[1,1]], dtype=float)
y = np.array([[0],[1],[1],[0]], dtype=float)

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -500, 500)))

def train_xor(lr, epochs=20000, seed=42, hidden=4):
    rng = np.random.default_rng(seed)
    W1 = rng.normal(0, 1, (2, hidden)); b1 = np.zeros((1, hidden))
    W2 = rng.normal(0, 1, (hidden, 1)); b2 = np.zeros((1, 1))
    loss = []
    for _ in range(epochs):
        h = sigmoid(X @ W1 + b1)
        o = sigmoid(h @ W2 + b2)
        loss.append(float(np.mean((o - y)**2)))
        do = (o - y) * o * (1.0 - o)
        dW2 = h.T @ do; db2 = do.sum(0, keepdims=True)
        dh = (do @ W2.T) * h * (1.0 - h)
        dW1 = X.T @ dh; db1 = dh.sum(0, keepdims=True)
        W2 -= lr * dW2; b2 -= lr * db2
        W1 -= lr * dW1; b1 -= lr * db1
    return (W1, b1, W2, b2), np.array(loss)

for lr in (0.05, 0.50, 2.00, 10.00):
    params, l = train_xor(lr)
    print(f"lr={lr:5.2f}: final MSE = {l[-1]:.2e}")
```
