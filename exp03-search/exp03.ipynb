```python
import random
import numpy as np
from collections import deque

def carve_maze(H=23, W=35, seed=11):
    r_gen = random.Random(seed)
    maze = np.ones((H, W), dtype=int)
    stack = [(1, 1)]; maze[1, 1] = 0
    while stack:
        r, c = stack[-1]; nbrs = []
        for dr, dc in [(-2,0),(2,0),(0,-2),(0,2)]:
            nr, nc = r+dr, c+dc
            if 1 <= nr < H-1 and 1 <= nc < W-1 and maze[nr, nc] == 1:
                nbrs.append((nr, nc, r+dr//2, c+dc//2))
        if nbrs:
            nr, nc, wr, wc = r_gen.choice(nbrs)
            maze[wr, wc] = 0; maze[nr, nc] = 0; stack.append((nr, nc))
        else:
            stack.pop()
    for _ in range(18):
        rr, cc = r_gen.randrange(2, H-2), r_gen.randrange(2, W-2)
        if maze[rr, cc] == 1 and ((maze[rr-1,cc]==0 and maze[rr+1,cc]==0) or (maze[rr,cc-1]==0 and maze[rr,cc+1]==0)):
            maze[rr, cc] = 0
    return maze

maze = carve_maze()
H, W = maze.shape
START, GOAL = (1, 1), (H-2, W-2)

def neighbours(cell):
    r, c = cell
    for dr, dc in ((-1,0),(1,0),(0,-1),(0,1)):
        nr, nc = r+dr, c+dc
        if 0 <= nr < H and 0 <= nc < W and maze[nr, nc] == 0:
            yield (nr, nc)

def search(start, goal, mode="bfs"):
    frontier = deque([start]); parent = {start: None}; order = {}
    while frontier:
        node = frontier.popleft() if mode == "bfs" else frontier.pop()
        if node in order: continue
        order[node] = len(order)
        if node == goal: break
        for nxt in neighbours(node):
            if nxt not in parent:
                parent[nxt] = node; frontier.append(nxt)
    path, cur = [], goal
    if goal in parent:
        while cur is not None: path.append(cur); cur = parent[cur]
    return order, path[::-1]

for m in ("bfs", "dfs"):
    o, p = search(START, GOAL, m)
    valid = all(maze[r,c]==0 for r,c in p) and all(abs(r1-r2)+abs(c1-c2)==1 for (r1,c1),(r2,c2) in zip(p, p[1:]))
    print(f"{m.upper()}: expanded={len(o)}, path_len={len(p)}, valid={valid}")
```
