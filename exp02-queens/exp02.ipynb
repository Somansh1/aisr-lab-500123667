```python
from itertools import permutations
import time

def conflicts(board):
    n = len(board); bad = 0
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                bad += 1
    return bad

def is_safe(board, col, row):
    for c in range(col):
        if board[c] == row or abs(board[c] - row) == abs(c - col):
            return False
    return True

def solve(n=8):
    sols, nodes = [], [0]
    def place(board, col):
        if col == n:
            sols.append(tuple(board)); return
        for row in range(n):
            nodes[0] += 1
            if is_safe(board, col, row):
                board[col] = row
                place(board, col + 1)
                board[col] = -1
    place([-1] * n, 0)
    return sols, nodes[0]

for n in range(4, 9):
    t0 = time.perf_counter(); s_bt, nd = solve(n); tb = time.perf_counter() - t0
    t0 = time.perf_counter(); s_gt = [p for p in permutations(range(n)) if conflicts(p) == 0]; tg = time.perf_counter() - t0
    assert len(s_bt) == len(s_gt)
    print(f"n={n}: sols={len(s_bt)}, nodes={nd}, bt={tb:.4f}s, gt={tg:.4f}s")
```
