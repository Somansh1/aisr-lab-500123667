```python
import numpy as np

H, W = 5, 7
WALLS = {(1,2), (2,2), (3,2), (1,4), (2,4)}
GOAL, PIT, START = (0, W-1), (2, W-1), (H-1, 0)
A = [(-1,0), (1,0), (0,-1), (0,1)]

def step(s, a):
    r, c = s; dr, dc = A[a]; nr, nc = r+dr, c+dc
    if not (0 <= nr < H and 0 <= nc < W) or (nr, nc) in WALLS:
        nr, nc = r, c
    ns = (nr, nc)
    if ns == GOAL: return ns, +10.0, True
    if ns == PIT: return ns, -10.0, True
    return ns, -0.1, False

def train(eps, episodes=600, alpha=0.5, gamma=0.95, seed=7):
    rng = np.random.default_rng(seed)
    Q = np.zeros((H, W, 4)); curve = []
    for _ in range(episodes):
        s, total = START, 0.0
        for _ in range(120):
            a = int(rng.integers(4)) if rng.random() < eps else int(np.argmax(Q[s[0], s[1]]))
            ns, r, done = step(s, a); total += r
            Q[s[0], s[1], a] += alpha * (r + gamma * np.max(Q[ns[0], ns[1]]) * (not done) - Q[s[0], s[1], a])
            s = ns
            if done: break
        curve.append(total)
    return Q, np.array(curve)

for eps in (0.01, 0.10, 0.20, 0.50):
    Q, c = train(eps)
    print(f"eps={eps:.2f}: mean return (last 100) = {c[-100:].mean():.2f}")
```
