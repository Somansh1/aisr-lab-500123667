```python
import random

def conflicts(board):
    n = len(board); bad = 0
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                bad += 1
    return bad

def hill_climb(n=8, seed=0):
    r = random.Random(seed)
    board = [r.randrange(n) for _ in range(n)]
    while True:
        cur = conflicts(board)
        if cur == 0: return True
        best, move = cur, None
        for c in range(n):
            old = board[c]
            for row in range(n):
                if row == old: continue
                board[c] = row; v = conflicts(board)
                if v < best: best, move = v, (c, row)
            board[c] = old
        if move is None: return False
        board[move[0]] = move[1]

def hill_climb_restarts(n=8, restarts=10, seed=0):
    return any(hill_climb(n, seed=seed*1000 + k) for k in range(restarts))

trials = 200
p = sum(hill_climb(8, seed=200+s) for s in range(trials)) / trials
print(f"Plain hill climbing: {100*p:.1f}% ({trials} trials)")
for k in (1, 2, 5, 10, 20):
    ok = sum(hill_climb_restarts(8, k, seed=5000+k*500+s) for s in range(trials)) / trials
    print(f"k={k:>2}: measured={100*ok:.1f}%, predicted={100*(1-(1-p)**k):.1f}%")
```
